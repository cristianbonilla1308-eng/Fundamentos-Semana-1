```python
def suma_total(lista):
    return sum(lista)

calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)

print(f"Tercera calificación: {calificaciones[2]}")

nueva1 = float(input("Nueva calificación 1: "))
nueva2 = float(input("Nueva calificación 2: "))

nueva_tupla = calificaciones + (nueva1, nueva2)
print(f"Nueva tupla: {nueva_tupla}")

lista_ordenada = list(nueva_tupla)
lista_ordenada.sort(reverse=True)
print(f"Lista ordenada (mayor a menor): {lista_ordenada}")

print(f"Suma total: {suma_total(lista_ordenada)}")
```

```python
def buscar_telefono(agenda, nombre):
    return agenda.get(nombre)

agenda = {
    "Ana": "555-0101",
    "Luis": "555-0102",
    "Mía": "555-0103"
}

nombre_nuevo = input("Nombre del nuevo contacto: ")
telefono_nuevo = input("Teléfono del nuevo contacto: ")
agenda[nombre_nuevo] = telefono_nuevo

print(f"Contactos registrados: {', '.join(agenda.keys())}")

nombre_buscar = input("Nombre a buscar: ")
telefono_encontrado = buscar_telefono(agenda, nombre_buscar)
print(f"El teléfono de {nombre_buscar} es: {telefono_encontrado}")
```

```python
votos = {"Rojo": 0, "Azul": 0, "Verde": 0}

for i in range(1, 6):
    color = input(f"Voto {i} - Elige un color (Rojo/Azul/Verde): ")
    if color in votos:
        votos[color] += 1
    else:
        print("Color no válido, no se contabilizará.")

ganador = max(votos, key=votos.get)

for color, total in votos.items():
    print(f"{color}: {total} voto(s)")

print(f"\nEl color ganador es: {ganador} con {votos[ganador]} voto(s)")
```

```python
def pedir_entero(mensaje):
    while True:
        try:
            valor = int(input(mensaje))
            return valor
        except ValueError:
            print("Error: Debes ingresar un número entero válido. Intenta de nuevo.")

num1 = pedir_entero("Primer número entero: ")
num2 = pedir_entero("Segundo número entero: ")

try:
    resultado = num1 / num2
    print(f"{num1} dividido entre {num2} es: {resultado}")
except ZeroDivisionError:
    print("Error: No es posible dividir entre cero. Ingresa un divisor distinto de 0.")
```

```python
def mostrar_tupla_ordenada():
    numeros = (7.5, 9.0, 8.0, 6.5, 10.0)
    lista_ordenada = sorted(numeros, reverse=True)
    print(f"Tupla original: {numeros}")
    print(f"Ordenada (mayor a menor): {lista_ordenada}")


def buscar_telefono(contactos):
    nombre = input("Nombre a buscar: ")
    telefono = contactos.get(nombre)
    if telefono:
        print(f"El teléfono de {nombre} es: {telefono}")
    else:
        print(f"El contacto '{nombre}' no existe en la agenda.")


def dividir_numeros():
    try:
        num1 = int(input("Primer número entero: "))
        num2 = int(input("Segundo número entero: "))
        resultado = num1 / num2
        print(f"{num1} dividido entre {num2} es: {resultado}")
    except ValueError:
        print("Error: Debes ingresar números enteros válidos.")
    except ZeroDivisionError:
        print("Error: No es posible dividir entre cero.")


def analizar_mensaje():
    def contar_palabras(texto):
        return len(texto.split())

    mensaje = input("Escribe un mensaje: ")
    print(f"Longitud del mensaje: {len(mensaje)}")
    print(f"En mayúsculas: {mensaje.upper()}")
    print(f"Texto reemplazado: {mensaje.replace('Python', 'programación')}")
    print(f"Palabras totales: {contar_palabras(mensaje)}")


def menu():
    contactos = {
        "Ana": "555-0101",
        "Luis": "555-0102",
        "Mía": "555-0103"
    }

    while True:
        print("\n--- MENÚ PRINCIPAL ---")
        print("1. Mostrar tupla de números ordenada")
        print("2. Buscar teléfono en contactos")
        print("3. Dividir dos números")
        print("4. Analizar un mensaje de texto")
        print("5. Salir")

        opcion = input("Elige una opción (1-5): ")

        if opcion == "1":
            mostrar_tupla_ordenada()
        elif opcion == "2":
            buscar_telefono(contactos)
        elif opcion == "3":
            dividir_numeros()
        elif opcion == "4":
            analizar_mensaje()
        elif opcion == "5":
            print("Programa finalizado. ¡Hasta luego!")
            break
        else:
            print("Opción no válida. Intenta de nuevo.")


menu()
```

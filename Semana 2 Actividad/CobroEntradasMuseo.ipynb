```python
#ACTIVIDAD 2 COBRO DE MUSEO
PRECIO_NINO = 0.00
PRECIO_MENOR = 30.00
PRECIO_MAYOR = 45.00

EDAD_LIMITE_NINO = 3
EDAD_LIMITE_MAYOR_EDAD = 18
EDAD_MINIMA_ADULTO_MAYOR = 60

DESCUENTO_ADULTO_MAYOR = 0.12
DESCUENTO_PROFESOR = 0.10
DESCUENTO_ESTUDIANTE = 0.10


def calcular_precio_base(edad):
    if edad < EDAD_LIMITE_NINO:
        return PRECIO_NINO
    elif edad < EDAD_LIMITE_MAYOR_EDAD:
        return PRECIO_MENOR
    else:
        return PRECIO_MAYOR


def calcular_descuento(edad, es_adulto_mayor, es_profesor, es_estudiante):


    if es_adulto_mayor and edad >= EDAD_MINIMA_ADULTO_MAYOR:
        return DESCUENTO_ADULTO_MAYOR, "Adulto mayor"

    elif es_profesor:
        return DESCUENTO_PROFESOR, "Profesor"

    elif es_estudiante:
        return DESCUENTO_ESTUDIANTE, "Estudiante"

    else:
        return 0.00, "Sin descuento"


def main():

  
    total_visitantes = int(input("¿Cuántos visitantes ingresarán? "))

    total_recaudado = 0.00

    print("\n===== MUSEO DE ANTROPOLOGÍA E HISTORIA =====")

   
    for visitante in range(1, total_visitantes + 1):

        print(f"\n--- Visitante {visitante} ---")

        edad = int(input("Edad: "))

        
        es_adulto_mayor = edad >= EDAD_MINIMA_ADULTO_MAYOR

        es_profesor = input("¿Es profesor? (s/n): ").lower() == "s"

        es_estudiante = input("¿Es estudiante? (s/n): ").lower() == "s"

        precio_base = calcular_precio_base(edad)

        descuento, tipo_descuento = calcular_descuento(
            edad,
            es_adulto_mayor,
            es_profesor,
            es_estudiante
        )

        monto_descuento = precio_base * descuento
        precio_final = precio_base - monto_descuento

        total_recaudado += precio_final

        print(f"Precio base: ${precio_base:.2f}")
        print(f"Descuento aplicado: {tipo_descuento}")
        print(f"Monto del descuento: ${monto_descuento:.2f}")
        print(f"Precio final: ${precio_final:.2f}")

    print("           RESUMEN DE LA COMPRA")
    print("============================================")
    print(f"Total de visitantes: {total_visitantes}")
    print(f"Total a pagar: ${total_recaudado:.2f}")
    print("============================================")


main()
```

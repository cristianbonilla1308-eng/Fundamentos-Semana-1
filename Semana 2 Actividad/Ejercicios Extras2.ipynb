#EXTRA 1

```python
total = 0

while True:
    costo = float(input("Costo del boleto: "))

    if costo == 0:
        print("Menor de 3 años, sin cargo")
        continue

    total += costo

    if total >= 100:
        print("Aforo alcanzado, se detiene el registro")
        break

print(f"Total acumulado: ${total:.2f}")
```

#EXTRA 2

```python
total_visitantes = int(input("Número de visitantes: "))

adultos = 0
suma_edades = 0

for i in range(1, total_visitantes + 1):
    edad = int(input(f"Edad del visitante {i}: "))

    suma_edades += edad

    if edad >= 18:
        adultos += 1

promedio = suma_edades / total_visitantes

print(f"Adultos: {adultos}")
print(f"Promedio de edad: {promedio:.2f}")
```

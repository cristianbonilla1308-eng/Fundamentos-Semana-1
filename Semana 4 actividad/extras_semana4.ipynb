## Extra 1: Suma de todos los elementos de una matriz

**Procedimiento y lógica:**

- `suma_matriz(matriz)` recorre la matriz con dos ciclos `for` anidados: el ciclo externo recorre cada fila (una lista) y el ciclo interno recorre cada elemento dentro de esa fila. En cada iteración se acumula el valor en la variable `total`, la cual se **retorna** al final con `return`.
- `imprimir_matriz(matriz)` recorre cada fila de la matriz y usa `" ".join(...)` para unir los elementos de la fila con espacios (sin corchetes ni comas) y los imprime con `print`. Esta función **no retorna ningún valor**, solo muestra información en pantalla.
- Finalmente se define la matriz `[[1, 2, 3], [4, 5, 6], [7, 8, 9]]` y se aplican ambas funciones.

```
def suma_matriz(matriz):
    total = 0
    for fila in matriz:
        for elemento in fila:
            total += elemento
    return total


def imprimir_matriz(matriz):
    for fila in matriz:
        print(" ".join(str(x) for x in fila))


matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
imprimir_matriz(matriz)
print(f"La suma de todos los elementos de la matriz es: {suma_matriz(matriz)}")
```

## Extra 2: Matriz con suma por fila

**Procedimiento y lógica:**

- `suma_fila(fila)` recibe una sola fila (lista) y la recorre con un ciclo `for`, acumulando los valores en `total`, el cual se **retorna** con `return`.
- `mostrar_tabla(matriz)` recorre cada fila de la matriz e imprime sus elementos separados por espacios, sin corchetes ni comas. Esta función **no retorna valor**.
- Se define la matriz `[[3, 1, 4], [1, 5, 9], [2, 6, 5]]`, se muestra con `mostrar_tabla` y luego se recorre la matriz con `enumerate` para calcular e imprimir la suma de cada fila usando `suma_fila`.

```
def suma_fila(fila):
    total = 0
    for elemento in fila:
        total += elemento
    return total


def mostrar_tabla(matriz):
    for fila in matriz:
        print(" ".join(str(x) for x in fila))


matriz = [[3, 1, 4], [1, 5, 9], [2, 6, 5]]
mostrar_tabla(matriz)
for i, fila in enumerate(matriz):
    print(f"Suma de la fila {i}: {suma_fila(fila)}")
```

## Extra 3: Producto de una columna sin usar el operador `*`

**Procedimiento y lógica:**

- Se define la matriz `[[1, 2, 3], [4, 5, 6], [7, 8, 9], [2, 4, 6]]` de 4 filas y 3 columnas.
- `multiplicar_columna(matriz, columna)` recorre cada fila de la matriz y toma el elemento de la columna indicada (`fila[columna]`). Como está prohibido usar `*`, la multiplicación se simula con **sumas repetidas**: por cada elemento `valor`, se suma el `producto` acumulado hasta el momento `valor` veces usando un ciclo `for _ in range(valor)`, y el resultado se convierte en el nuevo `producto`. Al terminar de recorrer todas las filas, la función **retorna** el producto final con `return`.
- Se solicita al usuario el número de columna (`0`, `1` o `2`) con `input()` y se muestra el resultado.

```
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9], [2, 4, 6]]


def multiplicar_columna(matriz, columna):
    producto = 1
    for fila in matriz:
        valor = fila[columna]
        resultado = 0
        for _ in range(valor):
            resultado += producto
        producto = resultado
    return producto


columna = int(input("Numero de columna: "))
print(f"El producto de la columna {columna} es: {multiplicar_columna(matriz, columna)}")
```

## Extra 4: Tabla de Pitágoras de tamaño variable

**Procedimiento y lógica:**

- `generar_tabla(tamano)` construye una matriz de `tamano x tamano` con dos ciclos `for` anidados (renglón `i` y columna `j`, ambos de 1 a `tamano`). Como no se puede usar `*`, cada producto `i x j` se calcula con **sumas repetidas**: se suma `i` un total de `j` veces dentro de un ciclo `for _ in range(j)`. La función **retorna** la matriz completa con `return`.
- `imprimir_tabla(tabla)` recorre cada fila de la tabla y la imprime usando `"\t".join(...)` para separar los valores con tabulador, sin corchetes ni comas. Esta función **no retorna valor**.
- Se pide al usuario un tamaño con `input()`; si está fuera del rango 2-5 se muestra un mensaje de error. Si es válido, se genera e imprime la tabla, y luego se piden un renglón y una columna para consultar el producto **directamente en la matriz** (`tabla[renglon - 1][columna - 1]`), sin usar el operador `*`.

```
def generar_tabla(tamano):
    tabla = []
    for i in range(1, tamano + 1):
        fila = []
        for j in range(1, tamano + 1):
            valor = 0
            for _ in range(j):
                valor += i
            fila.append(valor)
        tabla.append(fila)
    return tabla


def imprimir_tabla(tabla):
    for fila in tabla:
        print("\t".join(str(x) for x in fila))


tamano = int(input("Tamano de la tabla: "))
if tamano < 2 or tamano > 5:
    print("Error: el tamaño debe estar entre 2 y 5.")
else:
    tabla = generar_tabla(tamano)
    imprimir_tabla(tabla)
    renglon = int(input("Renglon (factor): "))
    columna = int(input("Columna (factor): "))
    producto = tabla[renglon - 1][columna - 1]
    print(f"El producto de {renglon} x {columna} es: {producto}")
```

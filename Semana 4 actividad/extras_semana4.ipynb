```python
def suma_matriz(matriz):
    total = 0
    for fila in matriz:
        for elemento in fila:
            total += elemento
    return total


def imprimir_matriz(matriz):
    for fila in matriz:
        linea = ""
        for elemento in fila:
            linea += str(elemento) + " "
        print(linea.strip())


# Aplicación de ambas funciones
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

imprimir_matriz(matriz)
print(f"La suma de todos los elementos de la matriz es: {suma_matriz(matriz)}")
```

```python
def suma_fila(fila):
    total = 0
    for elemento in fila:
        total += elemento
    return total


def mostrar_tabla(matriz):
    for fila in matriz:
        print(" ".join(str(x) for x in fila))


matriz = [[3, 1, 4], [1, 5, 9], [2, 6, 5]]
mostrar_tabla(matriz)
for i, fila in enumerate(matriz):
    print(f"Suma de la fila {i}: {suma_fila(fila)}")
```

```python
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9], [2, 4, 6]]


def multiplicar_columna(matriz, columna):
    producto = 1
    for fila in matriz:
        valor = fila[columna]
        resultado = 0
        for _ in range(valor):
            resultado += producto
        producto = resultado
    return producto


columna = int(input("Numero de columna: "))
print(f"El producto de la columna {columna} es: {multiplicar_columna(matriz, columna)}")
```

```python
def generar_tabla(tamano):
    tabla = []
    for i in range(1, tamano + 1):
        fila = []
        for j in range(1, tamano + 1):
            valor = 0
            for _ in range(j):
                valor += i
            fila.append(valor)
        tabla.append(fila)
    return tabla


def imprimir_tabla(tabla):
    for fila in tabla:
        print("\t".join(str(x) for x in fila))


tamano = int(input("Tamano de la tabla: "))
if tamano < 2 or tamano > 5:
    print("Error: el tamaño debe estar entre 2 y 5.")
else:
    tabla = generar_tabla(tamano)
    imprimir_tabla(tabla)
    renglon = int(input("Renglon (factor): "))
    columna = int(input("Columna (factor): "))
    producto = tabla[renglon - 1][columna - 1]
    print(f"El producto de {renglon} x {columna} es: {producto}")
```
